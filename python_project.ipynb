## Creating sales column, as there's no sales column.

## Key Performance Indicator

### 1. Total Revenue

```python
Total_sales = df1['sales'].sum()

print(f"Total Revenue:${Total_sales:,.2f}")
```

### 2.  Total Orders

```python
Total_orders = df1['order_id'].nunique()
print(f"Total_orders:{Total_orders:,}")
```

### 3. Average Order Value

```python
average_order_value = Total_sales/Total_orders
print(f"average_order_value:${average_order_value:.1f}")
```

## Monthly growth

```python
Month_Name=df2['date'].dt.month_name().unique()
print(Month_Name)
```

```python
# Merging df2 and df1
df1 = pd.merge(df1,df2[['order_id','date']],on ='order_id', how = 'left')
```

```python
Monthly_growth = df1.groupby(df1['date'].dt.to_period('M')).agg({'sales':'sum','order_id':'nunique'}).rename(columns={'order_id':'Total_orders'})
```

### 4.  Sales growth%

```python
Monthly_sales_growth = Monthly_growth['sales'].pct_change()*100
print(f"Monthly growth%{Monthly_sales_growth}")
Avg_monthly_growth = Monthly_sales_growth.mean()
print(f"On average the shop grows by {Avg_monthly_growth:.2f}% every month.")
```

### 5. Order growth%

```python
Monthly_order_growth = Monthly_growth["Total_orders"].pct_change()*100
print(f"Monthly growth%{Monthly_order_growth}")
Avg_monthly_growth = Monthly_order_growth.mean()
print(f"On average the shop grows by {Avg_monthly_growth:.2f}% every month.")
```

# **EXPLORATORY DATA ANALYSIS**

# **A) DATE & TIME (TREND ANALYSIS)**

# **1. What is the monthly sales trend ?**

```python
df1['Month']=df1['date'].dt.month_name()
Monthly_sales = df1.groupby(df1['date'].dt.to_period('M'))['sales'].sum()

Monthly_sales.plot(kind='line',marker='o',figsize=(8,5))
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('sales')
plt.grid(True)
plt.show()
```

## This business experience extreme volatility,it's unpredictable.There's a massive increase in sales  the month of July/August followed by a drop in sales in the month of September.
## - Highest sales - above $72,000 (July)

## - Lowest sales - $64,000 (October)

## - Low sales -  $64,100 (September)

# **2.Which days of the week drives the most sales ?**

```python
df1['day_of_week'] = df1['date'].dt.day_name()

day_order= ['Sunday','Monday','Tuesday','Wednesday','Thursday','Friday','Saturday']

day_sales = df1.groupby('day_of_week')['sales'].sum().reindex(day_order)

plt.figure(figsize=(8,5))
day_sales.plot(kind='bar')

plt.title('Total Sales by Days of the week',fontsize=14)
plt.xlabel('Day of the week',fontsize=12)
plt.ylabel('Total sales ($)',fontsize=12)
plt.show()
```

## Friday drives the most sales followed by a strong sales on Thursday and Saturday, while least sales were made on Sunday.

## Friday leads with almost $140,000,while sunday trailing with 90,000

# **3. Which month were the most sales made ?**

```python
df1['month'] = df1['date'].dt.month_name()
monthly_sales = df1.groupby('month')['sales'].sum().reindex([
    'January', 'February', 'March', 'April', 'May', 'June', 
    'July', 'August', 'September', 'October', 'November', 'December']).sort_values(ascending=False).head(6)


plt.figure(figsize=(8, 4))
monthly_sales.plot(kind='bar', color='green')
plt.title('Month with the most sales')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.xticks(rotation=45)
plt.show()
```

## July and May are the **Golden months** for sales. July sales is at the peak, with more than $70000. Good sales were also made in March, November January and April.

# **4. Which month were the least sales made ?**

```python
df1['month'] = df1['date'].dt.month_name()
monthly_sales = df1.groupby('month')['sales'].sum().reindex([
    'January', 'February', 'March', 'April', 'May', 'June', 
    'July', 'August', 'September', 'October', 'November', 'December']).sort_values(ascending=True).head(6)


monthly_sales.plot(kind='bar', color='red')
plt.title('Month with the least sales')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.xticks(rotation=45)
plt.show()
```

## THe least sales were made in October, September, December, February June and August.

## **5.What are the peak hour for orders ?**

```python
df1 = pd.merge(df1,df2[['order_id','time']],on ='order_id', how = 'left')
df1['hour'] = pd.to_datetime(df1['time'], format='%H:%M:%S').dt.hour
hourly_orders = df1.groupby('hour')['order_id'].nunique()

plt.figure(figsize=(8, 4))
sns.lineplot(x=hourly_orders.index, y=hourly_orders.values, marker='o', color='red')
plt.title('Peak Hours for Orders')
plt.xlabel('Hour of the Day (24hr)')
plt.ylabel('Number of Orders')
plt.grid(True)
plt.show()
```

## Sales peak during Lunch and dinner. The lunch peak is slightly higher than dinner. 
## Lunch peaks at 2,500 0rders arround 12:00 PM, while dinner peaks arround 2,300 orders arround 6:00 PM

# **B) GENERAL ANALYSIS**

## **1. How do sales differ accross different categories ?**

```python
df4 = df4.merge(df3[['pizza_id',  'pizza_type_id']], on='pizza_type_id')
```

```python
df1 = df1.merge(df4[[ 'pizza_id','category']], on='pizza_id',how='left')
```

```python
category_analysis = df1.groupby('category')['sales'].sum()

plt.figure(figsize=(10,5))
plt.pie(category_analysis,labels=category_analysis.index,autopct='%1.1f%%',startangle=180,
        colors = sns.color_palette('pastel'))
plt.title('Sales Across Different Categories')
plt.axis('equal')
plt.show()
```

## The menu is incredibly well-balanced. No single category dominates.
## **Classsic** leads slightly at 26.9%, while **Veggie** is the lowest but still significant at 23.7%

## **2. Which pizza type is mostly preferred by customers ?**

```python
# 1. Join Order Details and Pizzas using 'pizza_id'
# This connects the number sold (Quantity) to the ID (Pizza Type ID))

# 2. Group by pizza_type_id and sum the quantity
# This shows the total volume sold for each type
pizza_popularity = df1.groupby('pizza_id')['quantity'].sum().sort_values(ascending=False).head(10)

# 3. Visualization
plt.figure(figsize=(8, 4))
pizza_popularity.plot(kind='bar', color='gold', edgecolor='black')
plt.title('Top 10 Most Preferred Pizza Types (by Total Quantity)')
plt.xlabel('Pizza Type ')
plt.ylabel('Total Units Sold')
plt.xticks(rotation=45)
plt.show()
```

## Customers prefer **Big Meat_s** and **Thai Chicken_l** over all others. **Big Meat_s** is the top choice for individual lunch orders.

## **Big Meat_s** exceeds 1,750 units(the number of orders made)

## **3.Which pizza size is mostly preferred by customers ?**

```python
df1 = pd.merge(df1,df3[['pizza_id','size']],on= 'pizza_id',how = 'left')
```

```python
pizza_popularity = df1.groupby('size')['sales'].sum().sort_values(ascending=True)

# 3. Visualization
plt.figure(figsize=(8, 3))
pizza_popularity.plot(kind='bar', color='purple')
plt.title('Most Preferred Pizza  (by size)')
plt.xlabel('size')
plt.ylabel('Total sales')
plt.xticks(rotation=0)
plt.show()
```

## Customers have a massive preference for Large(L) pizzas followed by Medium(M). Extra_large(xl) and the Largest(xxl) are almost non-existent in the data.

## Large pizzas generate above $350,000.
